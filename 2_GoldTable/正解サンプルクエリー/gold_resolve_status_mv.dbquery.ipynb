```sql
CREATE OR REPLACE MATERIALIZED VIEW handson_catalog.databricks.gold_resolve_status_mv AS
SELECT
  date_format(resolution_date, 'yyyy-MM') AS resolution_month,
  resolution_status,
  COUNT(*) AS resolve_count
FROM
  handson_catalog.databricks.silver_inquiry_sales
WHERE
  resolution_date IS NOT NULL
GROUP BY
  date_format(resolution_date, 'yyyy-MM'),
  resolution_status
ORDER BY
  resolution_month ASC;

-- 作成したマテリアライズドビューの内容を確認
SELECT * FROM handson_catalog.databricks.gold_resolve_status_mv
ORDER BY resolution_month ASC, resolution_status;
```

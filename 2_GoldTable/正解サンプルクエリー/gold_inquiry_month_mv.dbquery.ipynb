```sql
-- マテリアライズドビューの作成
CREATE OR REPLACE MATERIALIZED VIEW handson_catalog.databricks.gold_inquiry_month_mv AS
SELECT
  DATE_TRUNC('month', inquiry_date) AS month_date,
  inquiry_type,
  COUNT(*) AS inquiry_count
FROM
  handson_catalog.databricks.silver_inquiry_sales
GROUP BY
  DATE_TRUNC('month', inquiry_date),
  inquiry_type;

-- 結果の表示
SELECT * FROM handson_catalog.databricks.gold_inquiry_month_mv
ORDER BY month_date,inquiry_type;
```

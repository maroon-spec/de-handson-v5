```sql
CREATE OR REPLACE MATERIALIZED VIEW handson_catalog.databricks.gold_resolve_date_mv AS
SELECT
  inquiry_id,
  inquiry_date,
  resolution_date,
  DATEDIFF(resolution_date, inquiry_date) AS days_to_resolve
FROM
  handson_catalog.databricks.silver_inquiry_sales
WHERE
  resolution_date IS NOT NULL;

SELECT * FROM handson_catalog.databricks.gold_resolve_date_mv;
```
